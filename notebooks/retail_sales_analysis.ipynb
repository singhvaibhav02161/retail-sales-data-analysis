# Retail Sales Analysis

Portfolio notebook covering data validation, KPI calculation, regional/category analysis and monthly trends.

```python
import pandas as pd
import matplotlib.pyplot as plt

DATA = 'data/retail_sales.csv'
df = pd.read_csv(DATA)
df.head()
```

```python
df.info()
df.describe(include='all')
```

```python
df['Revenue'] = df['Quantity'] * df['UnitPrice']
df['Profit'] = df['Revenue'] - df['Cost']
summary = df[['Revenue','Profit']].sum()
summary
```

```python
regional = df.groupby('Region', as_index=False)[['Revenue','Profit']].sum().sort_values('Revenue', ascending=False)
regional
```

```python
monthly = df.groupby('Month', as_index=False)[['Revenue','Profit']].sum()
monthly
```
